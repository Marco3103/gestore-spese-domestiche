# Gestore delle spese domestiche
Il notebook contiene tre celle, le quali contengono rispettivamente:  
1. l'importazione delle librerie necessarie
2. le funzioni chiamate durante l'esecuzione del programma
3. l'avvio del programma, con la scelta dell'operazione da effettuare.

Si possono quindi eseguire nell'ordine prestabilito.    

```python
import csv, os, datetime
```

```python
def check_if_file_exists():

  """
  Verifica che il file che registra le transazioni esista. Se non presente,
  lo crea con le intestazioni delle colonne.
  """

  if not os.path.exists("transaction_history.csv"):
    columns = ['data','descrizione','importo']
    with open("transaction_history.csv", "w") as f:
      csv_writer = csv.writer(f)
      csv_writer.writerow(columns)


def file_is_empty():

  """
  Controlla se sono state già inserite delle transazioni nel file.
  Questo controllo viene fatto sia quando si mostra il report mensile sia per
  il report delle top 10 transazioni.
  """

  with open("transaction_history.csv", "r") as f:
    csv_reader = csv.reader(f)
    # verifico prima che ci siano delle spese già registrate
    rows = sum(1 for row in csv_reader)
    if rows<=1:
      print("Non hai ancora inserito delle spese. Torno al menù operazioni, in modo da poter aggiungere una transazione. \n")
      return True
    return False


def print_transactions(transactions, max_length_description):

  """
  Stampa una lista di transazioni.
  """

  for row in transactions:
    print(f"{row[0]}  {row[1]}",' ' * (max_length_description-len(row[1])), row[2])
  print('-' * 45)


def add_transaction():

  """
  Chiede all'utente di inserire i dati per aggiungere una transazione, che verrà
  salvata in un file CSV.
  """

  while True:
    try:
      date = input("Inserisci la data della transazione, nel formato GG/MM/AAAA: ")
      date_formatted = datetime.datetime.strptime(date, "%d/%m/%Y")
      amount = float(input("Inserisci l'importo della transazione: "))
      description = input("Inserisci la descrizione della transazione: ")
    except ValueError:
      print("Non hai inserito la data o l'importo nel formato corretto, riprova")
      continue
    except:
      print("Errore, riprova")
      continue

    # salvo in un CSV
    check_if_file_exists()
    with open("transaction_history.csv", "a") as f:
      csv_writer = csv.writer(f)
      csv_writer.writerow([date_formatted.strftime("%d-%m-%Y"),description,amount])

    print("\n Transazione aggiunta! \n ")
    break


def show_report():

  """
  Visualizza il report di tutte le spese, ordinate cronologicamente per anno
  e mese.
  """

  check_if_file_exists()
  if file_is_empty():
    pass
  else:
    with open("transaction_history.csv", "r") as f:
      csv_reader = csv.reader(f)
      print("Di seguito il report delle spese registrate,ordinate per anno e mese: \n")
      columns = next(csv_reader)
      print("%s\t %s\t%s" % tuple(columns))
      print('-' * 45)

      """salvo i dati in una lista, in modo da poter ordinare per data e per
      riuscire a riutilizzarli. Calcolo la lunghezza della descrizione più
      lunga, in modo da poter poi stampare in modo più ordinato le colonne."""

      max_length_description = 0
      formatted_list = []
      for row in csv_reader:
        row[0] = datetime.datetime.strptime(row[0], "%d-%m-%Y")
        row[0] = datetime.datetime.strftime(row[0], "%Y-%m")
        max_length_description = max(max_length_description, len(row[1]))
        formatted_list.append(row)

      formatted_list = sorted(formatted_list)
      print_transactions(formatted_list, max_length_description)


def show_top_transactions():

  """
  Stampa le 10 transazioni con l'importo più alto, ordinandole da quella con
  importo più alto.
  """

  check_if_file_exists()
  if file_is_empty():
    pass
  else:
    with open("transaction_history.csv", "r") as f:
      csv_reader = csv.reader(f)
      print("Di seguito il report delle 10 transazioni con l'importo più elevato, in ordine decrescente: \n")
      columns = next(csv_reader)
      print("%s\t\t%s\t %s" % tuple(columns), "\n", "-" * 45)
      ordered_transactions = sorted(list(csv_reader), key=lambda x:float(x[2]), reverse=True)

    max_length_description = max(len(row[1]) for row in ordered_transactions)
    print_transactions(ordered_transactions, max_length_description)
```

```python
print("""Benvenuto nel gestore delle spese domestiche! Qui potrai tracciare le
tue spese, generare report mensili e identificare le spese più significative.""")

# menù
while True:

  print("""\n Qui di seguito le operazioni disponibili: \n
  1) Aggiungi una transazione \n
  2) Visualizza il report mensile \n
  3) Visualizza le top 10 transazioni \n
  4) Esci dal programma
   \n """)

  try:
    op_number = int(input("Inserisci il numero dell'operazione che vuoi eseguire: \n"))
    assert 1<=op_number<=4, "Hai inserito un numero non corretto, riprova. \n"
  except ValueError:
    print("Non hai digitato un numero, riprova")
    continue
  except AssertionError as e:
    print(e)
    continue
  except:
    print("Errore, riprova")
    continue

  if op_number==1:
    add_transaction()
  elif op_number == 2:
    show_report()
  elif op_number == 3:
    show_top_transactions()
  elif op_number == 4:
    print("Chiusura in corso...")
    break
```
